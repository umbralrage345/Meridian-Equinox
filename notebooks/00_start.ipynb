# Project Meridian: Mission Orientation

Welcome to Project Meridian! You are in command of an autonomous interstellar probe navigating across four distinct operational phases. Each act uses machine learning to solve a real-world mission challenge, from astronomical discovery to autonomous landing. Decisions made in early acts carry physical consequences into subsequent flight envelopes.

```python
# Setup and package installation
REPO_URL = "https://github.com/placeholder/meridian"

import os
import subprocess
import sys
from pathlib import Path

# Enable FAST mode by default to respect compute budgets (set to "0" to disable)
os.environ.setdefault("MERIDIAN_FAST", "1")

try:
    import meridian
except ImportError:
    if "google.colab" in sys.modules:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", f"git+{REPO_URL}"], check=True)
        import meridian
    else:
        repo_dir = Path.cwd()
        while repo_dir.parent != repo_dir and not (repo_dir / "meridian").exists():
            repo_dir = repo_dir.parent
        if (repo_dir / "meridian").exists() and str(repo_dir) not in sys.path:
            sys.path.insert(0, str(repo_dir))
        import meridian

fast_str = "ON" if os.environ.get("MERIDIAN_FAST") == "1" else "OFF"
backend_str = "torch" if getattr(meridian, "HAS_TORCH", False) else "sklearn"
print(f"Meridian v{meridian.__version__} | FAST mode: {fast_str} | Act 3 backend: {backend_str}")
```

```python
# Set your team identifier here
TEAM_ID = "team-alpha"

from meridian.state import load_state
from meridian.ui import briefing, dashboard, debrief
state = load_state(TEAM_ID)
```

```python
briefing(0)
```

### Pre-Flight Telemetry Check

Before departing, let us check probe subsystems and register telemetry communication channels. Our onboard computer monitors thruster fuel, hull integrity, and sensor noise floor. These resources will be depleted or degraded based on your flight performance in upcoming acts.

```python
# Initial probe diagnostics check
print('--- PROBE MERIDIAN PRE-FLIGHT SYSTEM CHECK ---')
print(f'Team Identifier   : {TEAM_ID}')
print(f'Propulsion Fuel   : {state.resources.fuel_pct:.1f}%')
print(f'Hull Integrity    : {state.resources.hull_health:.1f}%')
print(f'Sensor Noise Floor: σ = {state.resources.sensor_noise:.3f}')
print('Status: All telemetry transceivers nominal.')
```

```python
# Initial mission dashboard
dashboard(state)
```

```python
debrief(state, 0)
```
